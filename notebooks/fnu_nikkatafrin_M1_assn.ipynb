## Solution:    
             Here in this session we are gonna read all the data from our file and then try to seperate all the lists individually.for that we are following the below method:
         1.Reading the context of the file.
         2.Creating empty lists for all six components of the lists such as:
             a. diagnosis
             b. ages
             c. genders.
             d. chest_pains
             e. blood_pressure
             f. cholesterol
         3. after creating the empty list we are using a for loop to read all the content of the file.
         4. In the for block we are appending the empty lists with the respective index value of the data.
         5. Then finally we are printing the lists which we have created using the print function.

```python
with open('C:/Users/nikka/Downloads/M1_Data.txt') as csvfile:
    readCSV = csv.reader(csvfile)

    diagnosis, ages, genders, chest_pains, blood_pressure, cholesterol = ([] for row in range(6))
    for row1 in readCSV:
        diagnosis.append(row1[0])
        ages.append(int(row1[1]))
        genders.append(row1[2])
        chest_pains.append(row1[3])
        blood_pressure.append(int(row1[4]))
        cholesterol.append(int(row1[5]))
    print("diagnosis:", diagnosis)
    print("\n")
    print("Ages:", ages)
    print("\n")
    print("Genders:", genders)
    print("\n")
    print("Chest Pains:", chest_pains)
    print("\n")
    print("Blood Pressure:", blood_pressure)
    print("\n")
    print("Cholesterol:", cholesterol)
```
