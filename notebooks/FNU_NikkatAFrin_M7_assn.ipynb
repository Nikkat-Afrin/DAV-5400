### Task 1:
Using regular expressions, extract the Capacity and Attendance counts for each concert from the
unformatted text string shown above and store them in two separate Python list objects, i.e., one list containing the
Capacity values and one list containing the Attendance values.

#### Solution:
1. import the regex library 



```python
# load the re module
import re
```

```python
txt="JUNE:*****Black Stone Cherry---CAPACITY---:1500 -- $ATTENDANCE: 1,315--GATE:--$28,492;*****Lady Gaga ----CAPACITY---:25,000--- $ATTENDANCE: 24,368---GATE:--$461,956#;*****Paramore ----CAPACITY---:3000 ---$ATTENDANCE: 3,000 ---GATE:-$150,000;*****Rage Against the Machine---CAPACITY---:12000 ---$ATTENDAMCE: 10.782 ---GATE: --$724,087;*****BEYONCE---CAPACITY--:20000---$ATTENDANCE: 20,000—-GATE:$2,400,000*****"
txt
```

```python
# 1.1 extract all capacity numbers
capacity=[int (re.sub(',','', i)) for i in ( re.findall('(?<=--:)[\d\,]+', txt))]
#capacity= re.findall('(?<=--:)[\d\,]+', txt)
capacity
```

#### **Capacity regex explanation**
- **(?<=--:)[\d\,]+** - This expression is used to extract all the capacity values from the text. 
- **(?<=** is a look behind before the match function.
- **--:** these are the characters which we are looking around for get the capacity numbers.
- **[\d\,]** these are used to get the numbers from 0 to 9 and also ',' if it is there between the numbers.This expression is placed under a set.
- **+** This is used to match one or more preceding event.


```python
# 1.2 extract all attendance count 
attendance=[int (re.sub('\W','',i)) for i in ( re.findall('(?<=ATTENDANCE: |ATTENDAMCE: )[\d\,\.]+', txt))]
#attendance= re.findall('(?<=ATTENDANCE: |ATTENDAMCE: )[\d\,\.]+', txt)
attendance
```

#### **Attendance count regex explanation**
- **(?<=ATTENDANCE: |ATTENDAMCE: )[\d\,\.]+** - This expression is used to extract all the Attendance count from the text. 
> As there is a spelling mistake in the text given i used both the spellings of Attendance and Attendamce.

- **(?<=** is a look behind before the match function.
- **ATTENDANCE: |ATTENDAMCE:** these are the characters which we are looking around for get the Attendance count.
- **[\d\,\.]** these are used to get the numbers from 0 to 9 and also ',' if it is there between the numbers.This expression is placed under a set. And for one value it is 10.782 so for that purpose i used'.' for that to be included.
- **+** This is used to match one or more preceding event.


### Task 2:Using regular expressions, extract the names of each musical artist from the unformatted text string and store them in a Python list object. When complete, your list should contain the following entries:

```python
# 2 extract the names of each musical artist
artist= re.findall('(?<=\*)[\w\ ]+', txt)
artist
```
