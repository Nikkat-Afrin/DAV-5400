## Task 1:

### __Step 1__: Apply appropriate column heading to the second column

```python
# rename the columns 
data.rename( columns={'Category':'Oil_type'}, inplace=True )
data
```

###  __Step 2__: Now the __pd.melt()__ function can be used to convert the table from __wide__ to __long format__.

- Note the use of the __id_vars__ and __value_vars__ parameters


- Also note that we are defining the name of a new column to store the Oil suppliers names: '__OIl_suppliers__'


- And the new columns that will store the number of gallons of oil purchased and consumed for each month is labeled as '__Purchased__' and '__Consumed__'.


```python
# Melting from wide to long format, first 2 columns are used to 
ids = ["Month","Oil_type"]

# the remaining columns are treated as data values to be pivoted into long format
values = data.columns[2:]

# use pd.melt() to convert the wide table to long format
# Note the use of the id_vars and value_vars parameters
# also note that we are defining the name of a new column to store the city names: 'City'
# while the new column that will store the frequency counts of on time / departed flights
# is labeled as 'Value'

data = pd.melt(frame=data, id_vars=ids, value_vars=values, var_name='Oil_suppliers', value_name='Value')

data
```

We now have the data set in long format. 
But we still have a data problem:

- We have many 'NaN' values in the '__Month__' column as a result of the __melt()__ operation

### __Step 3__: Correctly fill the NaN items in the 'Month' column

- Based on the content of the original data frame, it appears that the 'NaN' values in the 'Month' column should be filled with the 'Month' value from the preceding row. We can use the 'forward fill' method within the __fillna()__ function to do this:

```python
data['Month'].fillna(method='ffill', inplace = True)
data
```

```python
data['Value']
```

```python
# Splitting the 'Value' column into two columns
data[['Purchased', 'Consumed']] = data['Value'].str.split(':', expand=True)
# Dropping the original 'Value' column
data.drop(columns=['Value'], inplace=True)
data
```

### The data is now in 'tidy' format:

- Each row is now comprised of a single observation (i.e., the Purchase of either a particular Oil supplier or Oil type of a particular month)

```python
data.dtypes
```

```python
# Convert Purchased and Consumed columns to numeric
data['Purchased'] = pd.to_numeric(data['Purchased'])
data['Consumed'] = pd.to_numeric(data['Consumed'])

print(data.dtypes) 
```

## Task 2:

### Step1:

- Here I am trying to group by the data by the combination of "Oil_suppliers" and "Oil_type" columns. Then, this grouped data can be used to calculate the sum of the "Purchased" and "Consumed" columns for each group using the .agg() method.

```python

# Group by Oil_suppliers and calculate the sum of Purchased and Consumed
grouped_data = data.groupby(["Oil_suppliers","Oil_type"]).agg({"Purchased": "sum", "Consumed": "sum"})

print(grouped_data)
```

### Step 2:
- The Remaining column gives the amount of oil left after the consumption and it can be calculated by subtracting the Purchased column with the consumption column.

```python
# Calculating the remaining oil with the help of Purchased and consumed column
grouped_data['Remaining'] = grouped_data['Purchased']-grouped_data['Consumed']

grouped_data
```

- The amount of oil remaining at the end of the chronological period for each pair of oil types/ brands can be seen in the Remaining column of the grouped_data data frame.

### Step 1: Question 2
- Here I am trying to group by the data by the combination of "Oil_suppliers" and "Oil_type" columns. Then, this grouped data can be used to calculate the sum of the "Consumed" columns for each group using the .agg() method.

```python

# Group by Oil_suppliers and calculate the sum of Purchased and Consumed
group_data = data.groupby(["Oil_type","Oil_suppliers"]).agg({ "Consumed": "sum"})

group_data.reset_index(inplace=True)
group_data
```

```python
group_data.columns
```

### Step 2:
- To get the highest amount of the oil consumed in both the oil types im again using the group by function.

```python
max_df = group_data.groupby(["Oil_type"]).agg({ "Consumed": "max"})
max_df.reset_index(inplace=True)
max_df
```

### Step 3:
- With the help of __merge()__ i was try to apply __inner join__ on the oil_type and consumed to the group_data to also get which oil_suppliers consume the most amount of Oil.

```python
final_df = pd.merge(group_data, max_df, how='inner', on=['Oil_type', 'Consumed'])
final_df
```

At the end of the chronological period, the most consumed brand of oil for each of the two separate types of oil is __Gulf__ and __Caltex__.

### Converting the long format into the wide format

- To convert the long format into the wide format im using the __pivot_table()__ function.This pivots the DataFrame data by specifying the row index as a multi-level index consisting of the columns "Month" and "Oil_type".

- The __columns__ parameter specifies the columns to be used for creating separate columns in the pivoted DataFrame, which are the unique values from the __Oil_suppliers__ column.

- The __values__ parameter indicates which columns' values will be used to fill the pivoted DataFrame. In this case, it includes both __Consumed__ and __Purchased__ columns.

-  After pivoting the DataFrame, __.reset_index()__ resets the index to make the columns "Month" and "Oil_type" regular columns again. The __inplace=True__ parameter ensures that the changes are made directly to the df_pivot DataFrame.

```python
# Pivot the DataFrame
df_pivot = data.pivot_table(index=[ 'Month','Oil_type'], columns='Oil_suppliers', values=['Consumed', 'Purchased'])

# Reset the index to make 'Month' and 'Oil_type' normal columns again
df_pivot.reset_index(inplace=True)
df_pivot
```

As we can see above that we got our desired wide format but there is still a major problem in the data frame that is the month column is not in a ordered format for this we have to catrgorize the month column and then we should set the order and perform the same processes which we are doing in the above to convert it into the wide_format.

```python
# Define the order of the months
month_order = ['Open', 'Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']

# Convert 'Month' to a categorical type with the categories ordered
data['Month'] = pd.Categorical(data['Month'], categories=month_order, ordered=True)

# Pivot the DataFrame
data_pivot = data.pivot_table(index=['Month', 'Oil_type'], columns='Oil_suppliers', values=['Purchased', 'Consumed'])

# Reset the index to make 'Month' and 'Oil_type' normal columns again
data_pivot.reset_index(inplace=True)

data_pivot
```

Here we are trying to change the columns of the purchased and consumed columns by using the __.reindex()__ method.

```python
# Assuming df is your DataFrame and you want to move the 'Purchased' columns before the 'Consumed' columns
new_columns_order = [('Month', ''), ('Oil_type', ''), 
                     ('Purchased', 'Caltex'), ('Purchased', 'Gulf'), ('Purchased', 'Mobil'), 
                     ('Consumed', 'Caltex'), ('Consumed', 'Gulf'), ('Consumed', 'Mobil')]

data_pivot = data_pivot.reindex(new_columns_order, axis=1)
```
