### Task 1:
1. (5 Points) Write a Python/NumPy code block that finds the distinct array items that exist in either array a or array b
but which do not appear within both arrays:
a = np.array([1, 2, 3, 2, 3, 4, 3, 4, 5, 6])
b = np.array([7, 2, 10, 2, 7, 4, 9, 4, 9, 8])
For example, if a 2 is found in both array a and array b, your output should not include a 2. Furthermore, the content of
the output should be comprised solely of unique data values.

### Solution:

#### 1.  Importing numpy library:
- For this task, we need to import the numpy library by using the import keyword and numpy as np.
- This bascially helps for us to use all the inbuilt numpy functions.
#### 2. arrays a and b:
- These are the predefined arrays given to us and for this task we are going to convert the arrrays into sets.
#### 3. set a and b:
- To convert the arrays into set we used set fucntion.
#### 4. '^' XOR operation:
- To find the unique values i used the XOR operation for this which will basically checks the two sets and gives us the distinct values.



```python
import numpy as np
a = np.array([1, 2, 3, 2, 3, 4, 3, 4, 5, 6])
b = np.array([7, 2, 10, 2, 7, 4, 9, 4, 9, 8])
print(a,b)

set_a = set(a)#converting arrays into sets
set_b = set(b)

unidue_data_values = np.array(list(set_a ^ set_b))#using XOR operation for comparing sets

print(unidue_data_values)
```

## solution method 2:
- Here i used the setdiff1d inbuilt fucntion along with the union and intersect1d which will gather all the union of two sets and intersection.
- In union all the elements will be subtracted with the common intersecting elements then we will get unique elements.

```python
unidue_data_values= np.setdiff1d(np.union1d(a, b), np.intersect1d(a, b))

print(unidue_data_values)
```

5. (8 Points) For this problem you will load a small data set to your online GitHub repository and then read it into a
NumPy 2D array. You will then use the content that NumPy array to answer a series of analytical questions.

```python
import pandas
import numpy as np
# suppress scientific notation
np.set_printoptions(suppress = True)
# read the data file from GitHub and convert to a NumPy ndarray object
# be sure to replace the ‘link_to_raw_copy of data file in your Github Repository’ placeholder shown below
# with a link to the “raw” version of the file you pushed to your online GitHub repository
nyc_water = pandas.read_csv('https://raw.githubusercontent.com/Nikkat-Afrin/DAV-5400/master/M3_Data.csv').to_numpy()
# verify you have an ndarray object
type(nyc_water)
```

```python
nyc_water
```

```python
# Extract the per capita column
per_capita_column = nyc_water[:, 3]

# Find the index of the maximum per capita value
max_per_capita_index = np.argmax(per_capita_column)

# Get the year with the maximum per capita
year_with_max_per_capita = nyc_water[max_per_capita_index, 0]

print("Year with the maximum per capita daily water consumption:", year_with_max_per_capita)
```

```python
# Calculate annual change in population and add a new column
population_column = nyc_water[:, 1]
annual_change_column = np.append([np.nan], np.diff(population_column))

# Add the new column to the array
nyc_water = np.column_stack((nyc_water, annual_change_column))

print("Updated NYC Water array with the annual change in population:")
print(nyc_water)
```

```python
# Extract the population column
population_column = nyc_water[:, 1]

# Calculate median and variance
median_population = np.median(population_column)
variance_population = np.var(population_column)

print("Median population of the city:", median_population)
print("Variance of the population of the city:", variance_population)
```

```python
# Extract the annual change column
annual_change_column = nyc_water[:, -1]

# Calculate the average annual change, ignoring 'N/A' values
average_annual_change = np.nanmean(annual_change_column)

print("Average annual change in the population of the city:", average_annual_change)
```
